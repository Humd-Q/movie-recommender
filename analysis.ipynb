```python
import pickle
from pathlib import Path
with Path('model.pkl').open('rb') as f:
    movies = pickle.load(f)
with Path('similarity.pkl').open('rb') as f:
    similarity = pickle.load(f)
print(f'{len(movies):,} movies; similarity matrix {similarity.shape}')
```


The function below shows the similarity score of movies for a given title, ranking how similar the top 10 movies are.

```python
def recommend(title, limit=10):
    matches = movies.index[movies.title == title]
    if not len(matches):
        raise ValueError(f'Movie not found: {title}')
    idx = int(matches[0])
    scores = similarity.getrow(idx).toarray().ravel()
    order = scores.argsort()[::-1]
    return movies.iloc[[i for i in order if i != idx][:limit]][['title']]

recommend('Avatar')
```
